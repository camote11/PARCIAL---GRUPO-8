## 1. Escenario a simular

Una plataforma de streaming tipo Disney+. El usuario se registra, elige un plan
(Básico con anuncios o Premium sin anuncios), explora el catálogo de películas y series,
reproduce contenido, lo califica, lo guarda en "Mi lista" (y puede eliminarlo), revisa o limpia su historial, ve su perfil y cambia su contraseña.
También puede cambiar de plan y calcular cuánto pagaría por varios meses
(con 10% de descuento al pagar 12 meses o más).

### Relaciones
- **Herencia:** `Pelicula` y `Serie` heredan de `Contenido`.
- **Agregación:** `Plataforma` contiene usuarios y contenido.
- **Asociación:** `Usuario` usa un `Plan` (se guarda en el atributo `plan`) y se relaciona con el contenido que ve y guarda.

## 5. Ejecución de prueba
estas celdas importan las clases reales y prueban cada funcionalidad.

### 5.1 Crear la plataforma y cargar el catálogo

````python
from plataforma import Plataforma
from contenido import cargar_datos

plataforma = Plataforma("Disney+")
cargar_datos(plataforma)
plataforma.mostrar_catalogo()
````

### 5.2 Polimorfismo: `mostrar_info()` en cada subclase

````python
catalogo = plataforma.get_catalogo()
print(catalogo[2].mostrar_info()) # Pelicula
print(catalogo[5].mostrar_info()) # Serie
````

### 5.3 Buscar por título y filtrar por género

````python
plataforma.mostrar_catalogo(plataforma.buscar_por_titulo("man"))
plataforma.mostrar_catalogo(plataforma.filtrar_por_genero("accion"))
````

### 5.4 Registrar usuarios e iniciar sesión

````python
print(plataforma.registrar_usuario("Alex", "alex@gmail.com", "1234", "Premium"))
print(plataforma.registrar_usuario("Francisco", "francisco@hotmail.com", "1234", "Basico"))   # correo inválido
print(plataforma.registrar_usuario("Danuska", "alex@gmail.com", "0000", "Basico"))   # correo repetido
print(plataforma.registrar_usuario("Melany", "melany@gmail.com", "1234", "Gold"))    # plan inexistente
print()
print("Login correcto:", plataforma.iniciar_sesion("alex@gmail.com", "1234").get_nombre())
print("Login con clave incorrecta:", plataforma.iniciar_sesion("alex@gmail.com", "9999"))
````

### 5.5 Reproducir

````python
alex = plataforma.iniciar_sesion("alex@gmail.com", "1234")
coco = plataforma.buscar_por_titulo("Coco")[0]

print(plataforma.reproducir(alex, "Coco"))
print("Agregar a Mi lista:", alex.agregar_a_mi_lista(coco))
print("Agregar de nuevo:", alex.agregar_a_mi_lista(coco), "(ya estaba)")

print("Mi lista:", [c.get_titulo() for c in alex.get_mi_lista()])
print("Historial:", [c.get_titulo() for c in alex.get_historial()])
````

### 5.6 Calificar

````python
coco = plataforma.buscar_por_titulo("Coco")[0]
print("Calificar con 5:", coco.calificar(5))
print("Calificar con 4:", coco.calificar(4))
print("Calificar con 9:", coco.calificar(9), "(inválido)")
print("Promedio de Coco:", coco.promedio())
print(coco.mostrar_info())
````

### 5.7 Planes: costo, cambio de plan

````python
plataforma.mostrar_planes()
print()
print("Plan actual:", alex.get_plan())
print("Total 12 meses (con descuento): $%.2f" % plataforma.calcular_costo(alex, 12))
print(plataforma.reproducir(alex, "Frozen"))

print()
print(plataforma.cambiar_plan(alex, "Basico"))
print(plataforma.cambiar_plan(alex, "Basico"))
print("Total 3 meses: $%.2f" % plataforma.calcular_costo(alex, 3))
print(plataforma.reproducir(alex, "Frozen"))
````

### 6.8 Perfil del usuario

````python
print(alex.mostrar_perfil())
````

### 5.9 Eliminar de Mi lista y limpiar historial

````python
print("Eliminar Coco:", alex.eliminar_de_mi_lista(coco))
print("Eliminar Coco otra vez:", alex.eliminar_de_mi_lista(coco), "(ya no estaba)")
print("Mi lista:", [c.get_titulo() for c in alex.get_mi_lista()])

print()
print("Historial antes:", len(alex.get_historial()), "título(s)")
alex.limpiar_historial()
print("Historial después:", len(alex.get_historial()), "título(s)")
````

### 6.10 Cambiar contraseña

````python
print("Con clave actual incorrecta:", alex.cambiar_clave("9999", "nueva123"))
print("Con clave actual correcta:", alex.cambiar_clave("1234", "nueva123"))
print("Login con clave vieja:", plataforma.iniciar_sesion("alex@gmail.com", "1234"))
print("Login con clave nueva:", plataforma.iniciar_sesion("alex@gmail.com", "nueva123").get_nombre())
````

## 6. Reglas de negocio
- El correo debe ser de un dominio común (gmail, hotmail, outlook, yahoo o icloud) y terminar en `.com`.
- Un correo no se puede registrar dos veces.
- El nombre y la clave no pueden estar vacíos.
- Las calificaciones solo aceptan puntajes del 1 al 5.
- El plan Básico muestra anuncios al reproducir; el Premium no.
- Pagar 12 meses o más da 10% de descuento.
- Para cambiar la contraseña hay que conocer la actual.
- Las búsquedas ignoran mayúsculas y tildes.
