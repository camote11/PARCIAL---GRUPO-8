## 1. Escenario a simular

Una plataforma de streaming tipo Disney+. El usuario se registra, elige un plan
(Básico con anuncios o Premium sin anuncios), explora el catálogo de películas y series,
reproduce contenido, lo califica, lo guarda en "Mi lista" (y puede eliminarlo), revisa o limpia su historial, ve su perfil y cambia su contraseña.
También puede cambiar de plan y calcular cuánto pagaría por varios meses
(con 10% de descuento al pagar 12 meses o más).

### Relaciones
- **Herencia:** `Pelicula` y `Serie` heredan de `Contenido`.
- **Agregación:** `Plataforma` contiene usuarios y contenido.
- **Asociación:** `Usuario` usa un `Plan` (se guarda en el atributo `plan`) y se relaciona con el contenido que ve y guarda.
